```
import pandas as pd

df = pd.read_csv(
    "/content/drive/MyDrive/US_Accidents_March23.csv",
    nrows=100000
)

df.head()
```

```
df = df.drop(columns=[
    "Description",
    "ID",
    "Source",
    "Zipcode",
    "Timezone",
    "Country",
    "Astronomical_Twilight",
    "Nautical_Twilight",
    "Civil_Twilight",
    "End_Lat",
    "End_Lng",
    "Turning_Loop"
], errors='ignore')
```

```
df.info()
```

```
#convert datatype of End_Time & Start_Time into datetime
df['Start_Time'] = pd.to_datetime(df['Start_Time'])
df['End_Time'] = pd.to_datetime(df['End_Time'])
```

```
df['start_year'] = df['Start_Time'].dt.year
df['start_month'] = df['Start_Time'].dt.month
df['start_day'] = df['Start_Time'].dt.day
df['start_hour'] = df['Start_Time'].dt.hour
```

```
df.info()
```

```
df.isnull().sum()
```

```
#for num cols use median to fill missing and median better than mean (sensitive with outliers)
num_cols = [
    'Temperature(F)',
    'Humidity(%)',
    'Pressure(in)',
    'Visibility(mi)',
    'Wind_Speed(mph)',
    'Wind_Chill(F)'
]

for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)

# precipitation usually = 0 when missing
df['Precipitation(in)'].fillna(0, inplace=True)
```

```
#for categorical cols use mode ,assume that the missing was the most frequent data
cat_cols = [
    'Weather_Condition',
    'Wind_Direction',
    'Sunrise_Sunset',
    'City'
]
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```
